```python
import torch

embeddings = torch.randn(4, 8)
```

```python
import torch.nn as nn

query = nn.Linear(8, 8)
key = nn.Linear(8, 8)
value = nn.Linear(8, 8)
```

```python
Q = query(embeddings)
K = key(embeddings)
V = value(embeddings)
scores = Q @ K.T
attention_weights = torch.softmax(scores, dim=1)
attention_output = attention_weights @ V
print(attention_output.shape)
```
