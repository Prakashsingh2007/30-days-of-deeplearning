```python
import torch
import torch.nn as nn
batch_size = 2
sequence_length = 6
vocab_size = 100

token_ids = torch.randint(0, vocab_size, (batch_size, sequence_length))

print("Token IDs Shape:", token_ids.shape)
```

```python
embedding = nn.Embedding(vocab_size, 32)

embeddings = embedding(token_ids)

print("Embedding Shape:", embeddings.shape)
```

```python
encoder_layer = nn.TransformerEncoderLayer(
    d_model=32,
    nhead=4,
    batch_first=True
)

encoder = nn.TransformerEncoder(
    encoder_layer,
    num_layers=2
)

encoder_output = encoder(embeddings)

print("Encoder Output Shape:", encoder_output.shape)
cls_output = encoder_output[:, 0, :]

print("First Token Shape:", cls_output.shape)
```

```python
classifier = nn.Sequential(
    nn.Linear(32, 16),
    nn.ReLU(),
    nn.Linear(16, 2)
)

logits = classifier(cls_output)

print("Final Logits Shape:", logits.shape)
```
