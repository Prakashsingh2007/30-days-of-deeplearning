```python
import torch
import torch.nn as nn
```

```python
vocab_size = 100
embedding_dim = 16

embedding = nn.Embedding(vocab_size, embedding_dim)

token_ids = torch.randint(0, vocab_size, (2, 5))

x = embedding(token_ids)

print("After Embedding:", x.shape)
```

```python
attention = nn.MultiheadAttention(
    embed_dim=embedding_dim,
    num_heads=4,
    batch_first=True
)

attn_output, _ = attention(x, x, x)

print("After Multihead Attention:", attn_output.shape)
```

```python
x = x + attn_output

print("After First Residual:", x.shape)
norm1 = nn.LayerNorm(embedding_dim)

x = norm1(x)

print("After First LayerNorm:", x.shape)
```

```python
ffn = nn.Sequential(
    nn.Linear(embedding_dim, 32),
    nn.ReLU(),
    nn.Linear(32, embedding_dim)
)

ffn_output = ffn(x)

print("After Feed Forward:", ffn_output.shape)
```

```python
x = x + ffn_output

print("After Second Residual:", x.shape)
norm2 = nn.LayerNorm(embedding_dim)

x = norm2(x)

print("After Second LayerNorm:", x.shape)
```
