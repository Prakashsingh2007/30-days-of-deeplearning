```python
import torch.nn as nn
from torchgen import model
model = nn.Sequential(
    nn.Linear(2, 2),
    nn.ReLU(),
    nn.Linear(2, 1)
)
x = torch.tensor([[1.0, 2.0]])
output = model(x)
print(output)
```

```python
model = nn.Sequential(
    nn.Linear(4, 16),
    nn.ReLU(),
    nn.Linear(16, 8),
    nn.ReLU(),
    nn.Linear(8, 3)
)
x = torch.randn(5, 4)  # Batch of 5 samples, each with 4 features
output = model(x)
print(output)
```
