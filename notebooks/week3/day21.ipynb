```python
import torch
import torch.nn as nn
```

```python
vocab = {
    "<PAD>": 0,
    "i": 1,
    "love": 2,
    "hate": 3,
    "ai": 4,
    "python": 5
}
```

```python
sentences = [
    "I love AI",
    "I hate Python"
]

encoded_sentences = []

for sentence in sentences:
    tokens = sentence.lower().split()
    encoded = [vocab[token] for token in tokens]
    encoded_sentences.append(encoded)

print("Encoded Input:")
print(encoded_sentences)
```

```python
inputs = torch.tensor(encoded_sentences)

print(inputs)
```

```python
class SentimentClassifier(nn.Module):

    def __init__(self):
        super().__init__()

        self.embedding = nn.Embedding(
            num_embeddings=6,
            embedding_dim=8
        )

        self.fc = nn.Linear(8, 2)

    def forward(self, x):

        embeddings = self.embedding(x)

        print("Embedding output shape:", embeddings.shape)

        pooled = embeddings.mean(dim=1)

        logits = self.fc(pooled)

        return logits
```

```python
model = SentimentClassifier()
logits = model(inputs)

print("Final logits shape:", logits.shape)

print(logits)
```
